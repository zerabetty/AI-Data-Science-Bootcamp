# Regression Models - Error Evaluation

This notebook demonstrates the fundamentals of simple linear regression and evaluates model performance using MSE, RMSE, and MAE.

The objective is to predict employee salaries based on years of experience using a predefined linear regression equation.

## 1. Problem Definition

Given employee experience and salary data, construct a simple linear regression model using the provided bias and weight values, generate salary predictions, and evaluate prediction errors.

```python
import pandas as pd
import numpy as np

experience = [5, 7, 3, 3, 2, 7, 3, 10, 6, 4, 8, 1, 1, 9, 1]
salary = [600, 900, 550, 500, 400, 950, 540, 1200, 900, 550, 1100, 460, 400, 1000, 380]

df = pd.DataFrame({
    "Experience": experience,
    "Salary": salary
})

df
```

## 2. Simple Linear Regression

A simple linear regression model predicts the target variable using a straight-line equation.

The regression equation is:

\[
\hat{y}=b+wx
\]

where:

- **ŷ** = Predicted salary
- **b** = Bias (Intercept)
- **w** = Weight (Slope)
- **x** = Years of experience

For this exercise:

- Bias = **275**
- Weight = **90**

## 3. Define the Regression Model

A simple linear regression model predicts the target variable using a straight-line equation.

For this exercise, the model parameters are predefined:

- Bias (Intercept) = 275
- Weight (Slope) = 90

The prediction equation is:

\[
\hat{y} = 275 + 90 \times x
\]

```python
# Define model parameters
bias = 275
weight = 90

# Calculate predicted salaries
df["Prediction"] = bias + weight * df["Experience"]

df.head()
```

## 3. Calculate Prediction Errors

The prediction error (also called the residual) is the difference between the actual salary and the predicted salary.

Three different error values will be calculated:

- Prediction Error
- Absolute Error
- Squared Error

```python
# Prediction Error (Residual)
df["Error"] = df["Salary"] - df["Prediction"]

# Absolute Error
df["Absolute_Error"] = df["Error"].abs()

# Squared Error
df["Squared_Error"] = df["Error"] ** 2

df
```

## 4. Regression Error Metrics

Evaluate the regression model using three common regression metrics:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)

```python
# Mean Absolute Error
mae = df["Absolute_Error"].mean()

# Mean Squared Error
mse = df["Squared_Error"].mean()

# Root Mean Squared Error
rmse = np.sqrt(mse)

print(f"MAE : {mae:.2f}")
print(f"MSE : {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
```

## 5. Interpretation

- **MAE** measures the average magnitude of prediction errors.
- **MSE** squares each error, giving greater importance to larger mistakes.
- **RMSE** is the square root of MSE and is expressed in the same unit as the target variable, making it easier to interpret.
